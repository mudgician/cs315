```python
def eval_2d_gauss(x,y,u_x,u_y,s_x,s_y):
    answ = 1.0/(2*np.pi*s_x*s_y)*np.exp(-0.5*(((x-u_x)/s_x)**2+((y-u_y)/s_y)**2))
    return answ
```
