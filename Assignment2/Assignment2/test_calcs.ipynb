```python
import numpy as np
from logistic_math import sigmoid
X = np.array([[1, 9, 5],[3, 3, 5],[0.7, 0.9, 0.2]])
w = np.array([1, 1, 1])
I = np.eye(3,3)
H = sigmoid(w.T@X)@(1-sigmoid(w.T@X))
H = np.sum(H*X@X.T) + I
print(H)
```
